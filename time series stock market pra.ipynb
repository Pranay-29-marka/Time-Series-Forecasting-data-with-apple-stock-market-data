```python
#Handiling the missing values 
df.fillna(method = 'ffill', inplace = True)
df.fillna(method = 'bfill', inplace = True)
df.head(20)
```

```python
#removing duplicates from dataset
df.dropna(axis = 1)
df.dropna(how = 'all')
df.dropna(thresh = 3)
df.dropna(subset = ['cci_7', 'cci_14'])#drop the rows in  a specific coloumns
df.dropna(inplace = True)
df
```

```python
#resampiling or summarize the dataset
#Monthly summary with Open, High, Low, Close, and Volume profits
monthly_open_trading = df['open'].resample('m').mean()#month profit
monthly_open_trading
quater_high_trading = df['high'].resample('q').max()#high price in last three months or quater month
quater_high_trading
monthly_close_trading = df['close'].resample('d').mean()#daily close profit in month
monthly_close_trading
monthly_low_trading = df['low'].resample('y').min()#lowest trading in month
monthly_low_trading
```

```python
#implementing the arima model
from statsmodels.tsa.arima.model import ARIMA 
from sklearn.metrics import mean_squared_error 
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Train-test split
Train_size = int(len(df) * 0.8)
Train_size
```

```python
#testing the first data
train, test = df['open'][:Train_size], df['open'][Train_size:]
```

```python
train,test
```

```python
#model fitting or training the train series
model = ARIMA(train, order=(5, 1, 0))  # p=5, d=1, q=0
model_fit = model.fit()
model_fit
```

```python
forecast = model_fit.forecast(steps=len(test))
forecast
df['close'][Train_size:]#predicting the nearest values
```

```python
#evalulating the model
arima_mse = mean_squared_error(test, forecast)
arima_mse
```

```python
ARIMA_RMSE =np.sqrt(arima_mse)
print("ARIMA RMSE:", ARIMA_RMSE)
```

```python
figure = plt.figure(figsize= (10, 6))
plt.plot(train.index, train, marker = '*', mec = 'g', mfc = 'k', label = 'training data')
plt.plot(test.index, test, marker = '*', label = 'actual or past data')
plt.plot(test.index, forecast, marker = '*', mec = 'k', mfc = 'k', label = 'predicting nearest or future values')
plt.legend(loc = 'upper left')
plt.grid(True)
```

```python
#SARIMA
"deals with indirectly trends & seasonal"
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
model_sarima = SARIMAX(train, order=(1,1,1), seasonal_order=(1,1,1,12))
model_fit_sarima = model_sarima.fit()
model_fit_sarima
```

```python
forecast_sarima = model_fit_sarima.forecast(steps = len(test))
forecast_sarima 
df['close'][:Train_size]

```

```python
df['close'][Train_size:]
```

```python
sarima_mse = mean_squared_error(test, forecast_sarima)
sarima_mse
```

```python
SARIMA_RMSE = np.sqrt(sarima_mse)
SARIMA_RMSE
SARIMA_RMSE_RMSE =np.sqrt(sarima_mse)
print("SARIMA RMSE:", SARIMA_RMSE)
```

```python
figure = plt.figure(figsize= (10, 6))
plt.plot(train.index, train, marker = '*', mec = 'g', mfc = 'k', label = 'training data')
plt.plot(test.index, test, marker = '*', label = 'actual or past data')
plt.plot(test.index, forecast_sarima, marker = '^', mec = 'k', mfc = 'k', label = 'predicting nearest or future values')
plt.legend(loc = 'upper left')
plt.grid(True)
```

```python
# -------- Create Final Dashboard Dataset --------
arima_df = pd.DataFrame({
    "Date": test.index,
    "ARIMA_Prediction": forecast
})

sarima_df = pd.DataFrame({
    "Date": test.index,
    "SARIMA_Prediction": forecast_sarima
})

```

```python
actual_df = pd.DataFrame({
    "Date": test.index,
    "Actual_Close": test.values
})

```

```python
final_output = actual_df.merge(arima_df, on="Date")
final_output = final_output.merge(sarima_df, on="Date")
```

```python
final_output["Best_Model"] = final_output.apply(
    lambda x: "ARIMA"
    if abs(x["Actual_Close"] - x["ARIMA_Prediction"])
       < abs(x["Actual_Close"] - x["SARIMA_Prediction"])
    else "SARIMA",
    axis=1
)
```
